```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
```

```python
load_dotenv()

client = OpenAI(
    api_key="YOUR_API_KEY"
)
```

```python
from PyPDF2 import PdfReader

pdf_path = r"C:\Users\Prajakta Mohane\Downloads\Prajakta_Mohane_Resume.pdf"

reader = PdfReader(pdf_path)

resume_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:  # Avoid None values
        resume_text += text

print("Resume Loaded Successfully!")
print(resume_text)
```

```python
prompt = f"""
Analyze this resume and provide:

1. Executive Summary

2. Top Skills

3. Strengths

4. Areas of Improvement

5. Missing Skills

6. Suitable Job Roles

7. Recommended Certifications

8. ATS Resume Suggestions

9. Keywords from Resume

10. 30-second Elevator Pitch

11. 3-month Career Roadmap

12. Salary Growth Suggestions

Resume:

{resume_text}
"""
```

```python
response = client.responses.create(
    model="gpt-5",
    input=prompt
)

analysis = response.output_text

print(analysis)
```

```python
job_role = input("Enter target job role: ")

response = client.responses.create(
    model="gpt-5",
    input=f"""
Write a professional cover letter for the role of {job_role}
using this resume.

Resume:

{resume_text}
"""
)

print(response.output_text)
```

```python
response = client.responses.create(
    model="gpt-5",
    input=f"""
Generate 10 LinkedIn headlines from this resume.

Resume:

{resume_text}
"""
)

print(response.output_text)
```

```python
system_prompt = f"""
You are an AI Resume Assistant.

Use the following resume to answer user questions.

Resume:
{resume_text}
"""

while True:

    question = input("\nAsk a question (type exit to stop): ")

    if question.lower() == "exit":
        break

    response = client.responses.create(
        model="gpt-5",
        input=system_prompt + "\nUser: " + question
    )

    print("\nAnswer:")
    print(response.output_text)
```
